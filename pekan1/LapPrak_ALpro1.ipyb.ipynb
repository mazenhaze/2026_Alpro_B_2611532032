# BAB II
# PEMBAHASAN

### 3.1 Python
Python adalah bahasa pemrograman *high-level*, ditafsirkan (*interpreted*), dan mendukung banyak paradigma pemrograman seperti prosedural, berorientasi objek (*object-oriented*), dan fungsional. Python dirancang dengan filosofi menekankan keterbacaan kode (*readability*) melalui penggunaan indentasi sebagai bagian dari sintaksis, bukan sekadar gaya penulisan. Karakteristik ini menjadikan Python populer digunakan dalam pengembangan aplikasi web, analisis data, kecerdasan buatan, otomasi, hingga pengembangan skrip sistem.

Sebagai bahasa yang bersifat *dynamically typed*, Python tidak mengharuskan pendeklarasian tipe data secara eksplisit, sehingga proses penulisan kode menjadi lebih cepat dan fleksibel. Python juga memiliki ekosistem pustaka (*library*) yang luas, seperti NumPy dan Pandas untuk pengolahan data, serta Django dan Flask untuk pengembangan web.

### 3.2 Visual Studio Code (VS Code)
Visual Studio Code adalah *source-code editor* ringan namun kaya fitur yang dikembangkan oleh Microsoft. VS Code mendukung berbagai bahasa pemrograman melalui mekanisme *extension*, termasuk dukungan penuh terhadap Python melalui ekstensi resmi *Python Extension for Visual Studio Code*. Fitur utama VS Code mencakup *syntax highlighting*, *IntelliSense* (pelengkapan kode otomatis), *debugging* terintegrasi, serta *integrated terminal* yang memungkinkan pengguna menjalankan perintah baris (*command line*) tanpa berpindah aplikasi.

Dalam konteks pengembangan berbasis Git, VS Code menyediakan antarmuka visual bawaan (*Source Control panel*) yang mempermudah pengguna memantau perubahan berkas, melakukan *staging*, *commit*, hingga sinkronisasi dengan repositori jarak jauh (*remote repository*) tanpa harus mengetik perintah Git secara manual, meskipun pemahaman terhadap perintah dasar Git tetap diperlukan.

### 3.3 Git dan GitHub
Git adalah sistem kontrol versi terdistribusi (*Distributed Version Control System*/DVCS) yang diciptakan oleh Linus Torvalds pada tahun 2005. Git memungkinkan pelacakan setiap perubahan pada berkas kode dari waktu ke waktu, memungkinkan pengembang untuk kembali ke versi sebelumnya, membuat cabang pengembangan (*branch*) secara paralel, serta menggabungkan (*merge*) hasil pekerjaan dari beberapa kontributor.

GitHub merupakan platform hosting berbasis web yang menyediakan layanan penyimpanan repositori Git secara daring, sekaligus dilengkapi fitur kolaborasi seperti *issue tracker*, *pull request*, dan *actions* untuk *Continuous Integration/Continuous Deployment* (CI/CD). Alur kerja umum yang digunakan meliputi:
- **Clone**: menyalin repositori dari GitHub ke komputer lokal.
- **Add/Stage**: menandai berkas yang akan disertakan dalam *commit*.
- **Commit**: menyimpan potongan perubahan (*snapshot*) beserta pesan deskriptif.
- **Push**: mengirimkan perubahan lokal ke repositori jarak jauh di GitHub.
- **Pull**: mengambil pembaruan terbaru dari repositori jarak jauh ke lokal.

### 3.4 Integrasi VS Code dengan GitHub
Koneksi antara VS Code dan GitHub umumnya dilakukan melalui dua metode utama, yaitu autentikasi menggunakan protokol HTTPS dengan *Personal Access Token* (PAT), atau menggunakan protokol SSH dengan pasangan kunci publik-privat (*public-private key pair*). Setelah autentikasi berhasil, VS Code dapat langsung berinteraksi dengan repositori GitHub melalui panel *Source Control* maupun ekstensi tambahan seperti *GitHub Pull Requests and Issues*, sehingga seluruh siklus pengembangan — mulai dari penulisan kode, pengujian, hingga publikasi perubahan — dapat dilakukan dalam satu lingkungan kerja terpadu.

### 2.2.1 Program "Hello World"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# program pertama
"""
komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print ("hello world")
print ()
print ("program ini menghasilkan")
print ("empat baris output")
print ("/ \\ // \\\\ /// \\\\\\")
print ("program ini" + " dibuat oleh a\n" + "M Abdurrauf_2611532032")
```

**Penjelasan :**
1. # program pertama
Ini adalah komentar satu baris, ditandai dengan tanda pagar (#). Baris ini tidak dieksekusi oleh Python, hanya digunakan sebagai catatan/keterangan bagi programmer, dalam hal ini menjelaskan bahwa ini adalah program pertama.

2. """
komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
Ini adalah komentar multi-baris (bisa juga disebut docstring), ditandai dengan tiga tanda kutip ("""..."""). Digunakan untuk menulis keterangan yang lebih panjang, dalam contoh ini menjelaskan konteks program yaitu untuk pertemuan pertama praktikum Algoritma Pemrograman (alpro).

3. print ("hello world")
Fungsi print() digunakan untuk menampilkan (mencetak) teks ke layar. Baris ini akan menghasilkan output: hello world.

4. print ()
Pemanggilan print() tanpa argumen (tanpa isi apa pun di dalam kurung) akan mencetak baris kosong, berfungsi sebagai pemisah antarbaris output agar tampilan lebih rapi.

5. print ("program ini menghasilkan")
print ("empat baris output")
Dua baris ini masing-masing mencetak teks sesuai isinya secara berurutan, 

6. print ("/ \\ // \\\\ /// \\\\\\")
Baris ini mencetak kombinasi garis miring. Perhatikan penggunaan karakter escape \\ yang merepresentasikan satu karakter backslash (\) karena tanda backslash tunggal biasanya digunakan Python untuk kombinasi karakter khusus (escape sequence), sehingga untuk mencetak backslash sungguhan harus ditulis dua kali. 

7. print ("program ini" + " dibuat oleh a\n" + "M Abdurrauf_2611532032")
Baris ini menggunakan operator + untuk menggabungkan (concatenate) beberapa string menjadi satu, lalu mencetaknya. Di dalamnya terdapat karakter escape \n yang berfungsi sebagai penanda baris baru (newline)
